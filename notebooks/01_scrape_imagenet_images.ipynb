```python
#code part 1
from bs4 import BeautifulSoup
import numpy as np
import requests
import cv2
import PIL.Image
import urllib

person_page = requests.get("http://www.image-net.org/api/text/imagenet.synset.geturls?wnid=n00007846")#person synset
person_filename=requests.get("http://www.image-net.org/api/text/imagenet.synset.geturls.getmapping?wnid=n00007846")
url = "http://www.image-net.org/api/text/imagenet.synset.geturls.getmapping?wnid=n00007846"
res = requests.get(url)
html_page = res.content
soup = BeautifulSoup(html_page, 'html.parser')
text = soup.find_all(text=True)
print(text)
str_soup=str(soup)#convert soup to string so it can be split
type(str_soup)
split_urls=str_soup.split('\r\n')#split so each url is a different possition on a list
#splited=str(split_urls).split(' ')
print(len(split_urls))
print (split_urls)
#print(splited)
j=0
person_image_name=[]
for i in split_urls:
  p1=str(i)
  p=p1.split()
  #print(len(p))
  k=0
  for l in p:
    if k==0:
      #print(l)
      #print(person_image_name)
      person_image_name.append(l)
      k=k+1
  j=j+1
print(person_image_name)  



print(person_page.content)

# BeautifulSoup is an HTML parsing library

person_soup = BeautifulSoup(person_page.content, 'html.parser')#puts the content of the website into the soup variable, each url on a different line
print(person_soup)
print(person_soup.prettify())
```
